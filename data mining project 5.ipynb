## Import Libraries

```python
import kagglehub
import pandas as pd
import os
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

## Download Dataset

```python
path = kagglehub.dataset_download("pavansubhasht/ibm-hr-analytics-attrition-dataset")
print(path)
```

## Read Dataset

```python
csv_file=os.path.join(path,"WA_Fn-UseC_-HR-Employee-Attrition.csv")
df=pd.read_csv(csv_file)
df.head()
```

## Shape

```python
print(df.shape)
```

## Info

```python
df.info()
```

## Missing

```python
print(df.isnull().sum())
```

## Encode

```python
le=LabelEncoder()
for c in df.columns:
    if df[c].dtype=="object":
        df[c]=le.fit_transform(df[c])
df.head()
```

## Split

```python
X=df.drop("Attrition",axis=1)
y=df["Attrition"]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

## Train

```python
model=RandomForestClassifier(n_estimators=100,random_state=42)
model.fit(X_train,y_train)
```

## Predict

```python
y_pred=model.predict(X_test)
print("Accuracy:",accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

## Importance

```python
imp=pd.DataFrame({"Feature":X.columns,"Importance":model.feature_importances_}).sort_values("Importance",ascending=False)
print(imp.head(10))
plt.figure(figsize=(10,6))
plt.bar(imp.head(10)["Feature"],imp.head(10)["Importance"])
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
